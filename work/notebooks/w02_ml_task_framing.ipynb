### Success metric

The primary technical metric will be the **Silhouette Score**, which measures how well-separated and internally consistent the clusters are.

However, a high Silhouette Score alone does not mean the clustering is useful. I will also evaluate whether the clusters are interpretable, reasonably sized, stable across reasonable choices of features or random seeds, and meaningfully different in their performance characteristics.

The final goal is therefore not simply to maximize a mathematical score, but to produce clusters that are both technically reasonable and useful for content decision-making.

The unit of analysis is one content page. Each row represents one content item, identified by `content_id`, with its associated search and performance measurements.

`content_id → performance features → cluster_id`

### Why ML beats a fixed rule

A fixed rule could manually define categories such as "high impressions" or "low CTR", but such thresholds may oversimplify the different ways pages can perform.

Clustering can consider several metrics together and discover combinations of characteristics that may not be obvious from individual thresholds. For example, two pages may have similar impressions but very different CTR, position, content age, or search demand.

The advantage of ML here is therefore not that it automatically decides what to do with a page. Its advantage is that it can help discover groups of pages with similar multi-dimensional performance patterns that can then be investigated and interpreted by a content team.
